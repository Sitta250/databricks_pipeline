```python
display(
    df_silver.filter(F.col("customer_name") != F.trim(F.col("customer_name")))
)
```

```python
df_silver = df_silver.withColumn(
    "customer_name",
    F.trim(F.col("customer_name"))
)
```

```python
display(
    df_silver.filter(F.col("customer_name") != F.trim(F.col("customer_name")))
)
```

```python
df_silver.select('city').distinct().show()
```

```python
city_mapping = {
    'Bengaluruu': 'Bengaluru',
    'Bengalore': 'Bengaluru',

    'Hyderabadd': 'Hyderabad',
    'Hyderbad': 'Hyderabad',

    'NewDehli': 'New Delhi',
    'NewDheli': 'New Delhi',
    'NewDehlee': 'New Delhi'
}
```

```python
allowed = ['Bengaluru', 'Hyderabad', 'New Delhi']
df_silver = (
    df_silver
    .replace(city_mapping, subset=["city"])
    .withColumn(
        "city",
        F.when(F.col("city").isNull(), None)
        .when(F.col("city").isin(allowed), F.col("city"))
        .otherwise(None)
    )
)
```

```python
df_silver.select('city').distinct().show()
```

```python
df_silver.select('customer_name').distinct().show()
```

```python
df_silver = df_silver.withColumn(
    "customer_name",
    F.when(F.col("customer_name").isNull(),None)
    .otherwise(F.initcap("customer_name"))
)
df_silver.select('customer_name').distinct().show()
```

```python
df_silver.filter(F.col("city").isNull()).show(truncate=False)
```

```python
null_customer_names = ['Sprintx Nutrition', 'Zenathlete Foods', 'Zenathlete Foods', 'Peak Performance Store', 'Primefuel Nutrition', 'Recovery Lane']
df_silver.filter(F.col("customer_name").isin(null_customer_names)).show(truncate=False)
```

```python
customer_city_fix = {
    789403: 'New Delhi',

    789420: 'Bengaluru',

    789422: 'New Delhi',

    789503: 'Hyderabad',

    789521: 'Hyderabad',

    789603: 'Hyderabad'
}

df_fix = spark.createDataFrame(
    [(k,v) for k,v in customer_city_fix.items()],
    ["customer_id", "fixed_city"]
)

display(df_fix)
```

```python
df_silver = (
    df_silver
    .join(df_fix, "customer_id", "left")
    .withColumn(
        "city",
        F.coalesce("city", "fixed_city")
    )
    .drop("fixed_city")
)

display(df_silver)
```

```python
df_silver.printSchema()
```

```python
df_fix.printSchema()
```

```python
df_silver = df_silver.withColumn("customer_id", F.col("customer_id").cast("string"))
print(df_silver.printSchema())
```

```python
df_silver = (
    df_silver
    .withColumn(
        "customer",
        F.concat_ws("-", "customer_name", F.coalesce(F.col("city"), F.lit("Unknown")))
    )

    .withColumn("market", F.lit("India"))
    .withColumn("platform", F.lit("Sports Bar"))
    .withColumn("channel", F.lit("Acquisition"))
)

display(df_silver.limit(5))
```

```python
df_silver.write \
    .format("delta") \
    .option("mergeSchema", "true") \
    .mode("overwrite") \
    .saveAsTable(f"{catalog}.{silver_schema}.{data_source}")

spark.sql(f"""
ALTER TABLE {catalog}.{silver_schema}.{data_source}
SET TBLPROPERTIES (
  delta.enableChangeDataFeed = true
)
""")
```

## Gold Processing

```python
df_silver = spark.sql(f"SELECT * FROM {catalog}.{silver_schema}.{data_source};")

df_gold = df_silver.select("customer_id", "customer_name", "city", "customer", "market", "platform", "channel")
                      
            
```

```python
display(df_gold.limit(10))
```

```python
df_gold.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(f"{catalog}.{gold_schema}.sb_dim_{data_source}")

spark.sql(f"""
ALTER TABLE {catalog}.{gold_schema}.sb_dim_{data_source}
SET TBLPROPERTIES (
  delta.enableChangeDataFeed = true
)
""")
```

```python
delta_table = DeltaTable.forName(spark, "fmcg.gold.dim_customers")
df_child_customers = spark.table("fmcg.gold.sb_dim_customers").select(
    F.col("customer_id").alias("customer_code"),
    "customer",
    "market",
    "platform",
    "channel"
)
```

```python
delta_table.alias("target").merge(
    source=df_child_customers.alias("source"),
    condition="target.customer_code = source.customer_code"
).whenMatchedUpdateAll().whenNotMatchedInsertAll().execute()
```
